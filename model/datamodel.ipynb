```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
```

```python
n_total = 1500
positive_ratio = 0.66
n_positive = int(n_total * positive_ratio)
n_negative = n_total - n_positive
```

```python
def generate_positive(n):
    return pd.DataFrame({
        "Gender": np.random.choice(["Male", "Female"], n),
        "Age": np.random.randint(5, 80, n),
        "Haemoglobin": np.round(np.random.normal(11.5, 1.5, n), 1),
        "ESR": np.random.randint(25, 70, n),
        "WBC": np.round(np.random.normal(4.0, 1.2, n), 2),  # leukopenia common
        "Neutrophil": np.random.randint(40, 70, n),
        "Lymphocyte": np.random.randint(20, 50, n),
        "Monocyte": np.random.randint(5, 12, n),
        "Eosinophil": np.random.randint(0, 5, n),
        "Basophil": np.random.randint(0, 2, n),
        "RBC": np.round(np.random.normal(4.5, 0.5, n), 2),
        "Platelets": np.random.randint(10, 120, n),  # thrombocytopenia
        "Result": "Positive"
    })
```

```python
def generate_negative(n):
    return pd.DataFrame({
        "Gender": np.random.choice(["Male", "Female"], n),
        "Age": np.random.randint(5, 80, n),
        "Haemoglobin": np.round(np.random.normal(13.5, 1.2, n), 1),
        "ESR": np.random.randint(5, 25, n),
        "WBC": np.round(np.random.normal(7.5, 1.5, n), 2),
        "Neutrophil": np.random.randint(50, 75, n),
        "Lymphocyte": np.random.randint(15, 40, n),
        "Monocyte": np.random.randint(3, 8, n),
        "Eosinophil": np.random.randint(0, 4, n),
        "Basophil": np.random.randint(0, 2, n),
        "RBC": np.round(np.random.normal(5.0, 0.4, n), 2),
        "Platelets": np.random.randint(150, 400, n),
        "Result": "Negative"
    })
```

```python
df_positive = generate_positive(n_positive)
df_negative = generate_negative(n_negative)
```

```python
synthetic_cbc = pd.concat([df_positive, df_negative], ignore_index=True)
synthetic_cbc = synthetic_cbc.sample(frac=1).reset_index(drop=True)
```

```python
file_path = "cbc_1500.csv"
synthetic_cbc.to_csv(file_path, index=False)

file_path
```
